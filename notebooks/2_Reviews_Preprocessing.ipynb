Make a copy

```python
data = df.copy()

print("Working dataset shape:", data.shape)
```

Check duplicates

```python
print("Duplicate review IDs:", data['review_id'].duplicated().sum())

print("Duplicate review texts:", data['review_text'].duplicated().sum())
```

Remove Duplicate review_ids

```python
data = data.drop_duplicates(subset='review_id', keep='first').reset_index(drop=True)

print("Shape after removing duplicate review IDs:", data.shape)
```

Handle missing review text

```python
before = len(data)

data = data.dropna(subset=['review_text']).copy()

data = data[data['review_text'].str.strip() != ''].copy()

data = data.reset_index(drop=True)

print("Rows before:", before)
print("Rows after:", len(data))
print("Reviews removed:", before - len(data))
```

Clean the review text

```python
def clean_text(text):
    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r'http\S+|www\S+|https\S+', ' ', text)

    # Remove HTML tags
    text = re.sub(r'<.*?>', ' ', text)

    # Remove email addresses
    text = re.sub(r'\S+@\S+', ' ', text)

    # Keep letters, numbers and basic punctuation
    text = re.sub(r'[^a-z0-9\s.,!?\'-]', ' ', text)

    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()

    return text


data['clean_review_text'] = data['review_text'].apply(clean_text)

display(
    data[['review_text', 'clean_review_text']].head(10)
)
```

Remove extremely short reviews

```python
data['word_count'] = data['clean_review_text'].str.split().str.len()

print(data['word_count'].describe())
```

```python
data = data[data['word_count'] >= 3].copy()

data.reset_index(drop=True, inplace=True)

print("Dataset after text filtering:", data.shape)
```

Normalize the score column

```python
print(data['score'].value_counts(dropna=False).head(30))
```

```python
def normalize_score(score):
    if pd.isna(score):
        return np.nan

    score = str(score).strip().lower()

    # "10 / 10.0", "9.8 / 10.0", etc.
    match = re.search(r'(\d+(?:\.\d+)?)\s*/\s*(\d+(?:\.\d+)?)', score)

    if match:
        value = float(match.group(1))
        maximum = float(match.group(2))

        if maximum > 0:
            return (value / maximum) * 100

    # "5 stars"
    match = re.search(r'(\d+(?:\.\d+)?)\s*stars?', score)

    if match:
        value = float(match.group(1))
        return (value / 5) * 100

    # Pure numeric value
    if re.fullmatch(r'\d+(?:\.\d+)?', score):
        value = float(score)

        if value <= 10:
            return value * 10

        return value

    # Text-based score
    if 'essential' in score:
        return 100

    return np.nan


data['score_normalized'] = data['score'].apply(normalize_score)

display(
    data[['score', 'score_normalized']].head(30)
)
```

Create sentiment labels

```python
def create_sentiment(score):
    if pd.isna(score):
        return np.nan

    if score >= 80:
        return 'positive'
    elif score >= 60:
        return 'neutral'
    else:
        return 'negative'


data['sentiment'] = data['score_normalized'].apply(create_sentiment)
```

```python
print(data['sentiment'].value_counts(dropna=False))

print("\nPercentage:")
print(
    data['sentiment']
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
)
```

Check the cleaned dataset

Now create the 77.5 / 20 / 2.5 split

```python
games = data['game_name'].dropna().unique()

print("Number of unique games:", len(games))
```

Split the games

```python
from sklearn.model_selection import train_test_split

# First: 77.5% training and 22.5% remaining
train_games, remaining_games = train_test_split(
    games,
    test_size=0.225,
    random_state=42
)

# From remaining 22.5%:
# 20% of total = 88.888...% of remaining
# 2.5% of total = 11.111...% of remaining

test_games, unseen_games = train_test_split(
    remaining_games,
    test_size=0.1111111111,
    random_state=42
)

print("Training games:", len(train_games))
print("Testing games:", len(test_games))
print("Unseen games:", len(unseen_games))

print("\nTotal:", len(train_games) + len(test_games) + len(unseen_games))
```

Create the three datasets

```python
train_df = data[data['game_name'].isin(train_games)].copy()

test_df = data[data['game_name'].isin(test_games)].copy()

unseen_df = data[data['game_name'].isin(unseen_games)].copy()
```

```python
print("TRAINING")
print(train_df.shape)

print("\nTESTING")
print(test_df.shape)

print("\nUNSEEN")
print(unseen_df.shape)
```

Verify there is no game overlap

```python
train_set = set(train_games)
test_set = set(test_games)
unseen_set = set(unseen_games)

print("Train ∩ Test:", len(train_set & test_set))
print("Train ∩ Unseen:", len(train_set & unseen_set))
print("Test ∩ Unseen:", len(test_set & unseen_set))
```

```python
assert len(train_set & test_set) == 0
assert len(train_set & unseen_set) == 0
assert len(test_set & unseen_set) == 0

print("No game overlap between train, test and unseen datasets.")
```
